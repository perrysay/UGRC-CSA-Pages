# 4.2 for Loops

## 1. What Is a For Loop?

A **for loop** repeats code while a condition is true. It is especially useful when you know how many times you want to repeat something.

Basic structure:

```java
for (initialization; condition; update) {
    // code to repeat
}
```

The three parts are:

1. **Initialization** – creates or sets the loop variable.
2. **Condition** – determines whether the loop should continue.
3. **Update** – changes the loop variable after each repetition.

````python
for (int i = 0; i < 5; i++) {
    System.out.println(i);
}
````

### What happens?

The loop prints:

```text
0
1
2
3
4
```

Notice that `i` starts at `0`, and the loop stops before `i` reaches `5`.

## 2. Understanding the Three Parts

Consider:

```java
for (int i = 1; i <= 5; i++) {
    System.out.println(i);
}
```

| Part | Code | Purpose |
|---|---|---|
| Initialization | `int i = 1` | Starts `i` at 1 |
| Condition | `i <= 5` | Continue while `i` is 5 or less |
| Update | `i++` | Increase `i` by 1 |

`i++` is shorthand for:

```java
i = i + 1;
```

## 3. Tracing a For Loop

Use a trace table to understand exactly what happens during each iteration.

```java
for (int i = 2; i <= 8; i += 2) {
    System.out.println(i);
}
```

| Iteration | `i` | `i <= 8` | Printed |
|---:|---:|:---:|---:|
| 1 | 2 | true | 2 |
| 2 | 4 | true | 4 |
| 3 | 6 | true | 6 |
| 4 | 8 | true | 8 |
| 5 | 10 | false | — |

The update `i += 2` increases `i` by 2 each time.

````python
for (int i = 2; i <= 8; i += 2) {
    System.out.println(i);
}
````

## 4. Counting Backward

A `for` loop does not have to count upward. You can decrease the loop variable.

```java
for (int i = 5; i >= 1; i--) {
    System.out.println(i);
}
```

Output:

```text
5
4
3
2
1
```

`i--` is shorthand for `i = i - 1`.

## 5. For Loops with Calculations

Loops are useful for performing calculations repeatedly.

Example: calculate the squares from 1 through 5.

````python
for (int i = 1; i <= 5; i++) {
    int square = i * i;
    System.out.println(square);
}
````

## 6. Using a For Loop with an Accumulator

An **accumulator** is a variable that stores a running total.

Example: find the sum of the numbers 1 through 10.

````python
int sum = 0;

for (int i = 1; i <= 10; i++) {
    sum += i;
}

System.out.println(sum);
````

The value of `sum` changes each time through the loop:

`0 → 1 → 3 → 6 → 10 → ... → 55`

The final answer is **55**.

## 7. Looping Through an Array

A `for` loop can access every element of an array using its index.

Remember that Java arrays start at index `0`.

````python
int[] scores = {90, 85, 92, 78, 88};

for (int i = 0; i < scores.length; i++) {
    System.out.println(scores[i]);
}
````

### Why use `i < scores.length`?

If an array has 5 elements, its indexes are `0, 1, 2, 3, 4`.

Therefore, the last valid index is `scores.length - 1`.

Using `i < scores.length` prevents an `ArrayIndexOutOfBoundsException`*.

*(`ArrayIndexOutOfBoundsException` is a Java error that happens when you try to access an array using an index that doesn't exist.)

## 8. Nested For Loops

A **nested loop** is a loop inside another loop.

Example:

````python
for (int row = 1; row <= 3; row++) {
    for (int col = 1; col <= 4; col++) {
        System.out.print("* ");
    }
    System.out.println();
}
````

Output:

```text
* * * *
* * * *
* * * *
```

The inner loop runs completely for every iteration of the outer loop.

## 9. Common Errors

### Error 1: Using `<=` when `<` is needed

```java
for (int i = 0; i <= scores.length; i++)
```

This tries to access an index that does not exist when used with `scores[i]`.

Use:

```java
for (int i = 0; i < scores.length; i++)
```

### Error 2: Forgetting the update

```java
for (int i = 0; i < 10; ) {
    System.out.println(i);
}
```

The value of `i` never changes, so the condition remains true. This can create an infinite loop.

### Error 3: Off-by-one errors

Pay close attention to whether your condition uses `<`, `<=`, `>`, or `>=`.

# AP Computer Science A Connection

For loops are commonly used in **iteration** problems on the AP Computer Science A exam.

When reading a loop, ask:

1. What is the starting value?
2. What condition must be true for another iteration to occur?
3. How does the loop variable change?
4. How many times does the loop execute?
5. What happens during each iteration?

# Quick Review

- `i++` increases `i` by 1.
- `i--` decreases `i` by 1.
- `i += 2` increases `i` by 2.
