## 1. What Is an Algorithm?

An **algorithm** is a step-by-step set of instructions used to solve a problem or complete a task.

For example, an algorithm that determines whether a student passes could be:
1. Get the student's score.
2. Check whether the score is at least 70.
3. If it is, report `Pass`.
4. Otherwise, report `Fail`.

Algorithms often use **selection** and **repetition** to control which instructions execute and how many times they execute.

## 2. Selection

**Selection** allows an algorithm to make a decision. The program evaluates a Boolean condition and chooses which block of code to execute.

Java commonly uses:
- `if` — executes a block when a condition is true.
- `else if` — checks another condition when an earlier condition was false.
- `else` — executes when none of the earlier conditions are true.

### Java Example
```java
int score = 85;

if (score >= 70) {
    System.out.println("Pass");
} else {
    System.out.println("Fail");
}
```

Since 85 is greater than or equal to 70, the program prints `Pass`.

**Key idea:** Selection means **choose a path based on a condition**.

## 3. Selection with Multiple Conditions

Selection can have more than two possible paths.

```java
int score = 85;

if (score >= 90) {
    System.out.println("A");
} else if (score >= 80) {
    System.out.println("B");
} else if (score >= 70) {
    System.out.println("C");
} else {
    System.out.println("F");
}
```

The conditions are checked from top to bottom. Once a condition is true, its block runs and the remaining `else if`/`else` blocks are skipped.

## 4. Repetition

**Repetition** allows an algorithm to execute instructions multiple times. This is also called **iteration**.

Two common Java loops are `for` and `while`.

### `for` Loop
A `for` loop is useful when the number of repetitions is known or controlled by a counter.

```java
for (int i = 0; i < 5; i++) {
    System.out.println("Hello");
}
```

This prints `Hello` five times.

### `while` Loop
A `while` loop repeats as long as its condition is true.

```java
int x = 0;

while (x < 5) {
    System.out.println(x);
    x++;
}
```

The loop stops when `x < 5` becomes false.

**Key idea:** Repetition means **repeat instructions**.

## 5. Understanding a Java `for` Loop

Look at this structure:

```java
for (int i = 0; i < 5; i++) {
    // repeated code
}
```

It has three parts:

| Part | Example | Purpose |
|---|---|---|
| Initialization | `int i = 0` | Creates the counter |
| Condition | `i < 5` | Determines whether the loop continues |
| Update | `i++` | Changes the counter after each iteration |

The values of `i` are `0, 1, 2, 3, 4`, so the loop runs **5 times**.

## 6. Selection vs. Repetition

| Structure | Purpose | Common Java statements |
|---|---|---|
| **Selection** | Makes a decision | `if`, `else if`, `else` |
| **Repetition** | Repeats instructions | `for`, `while` |

### Easy way to remember
- **Selection:** "Which path should I take?"
- **Repetition:** "How many times should I do this?"

## 7. Combining Selection and Repetition

Algorithms often combine both structures. For example, we can examine several scores and determine whether each one is a passing score.

```java
int[] scores = {85, 62, 91, 74, 55};

for (int score : scores) {
    if (score >= 70) {
        System.out.println(score + " Pass");
    } else {
        System.out.println(score + " Fail");
    }
}
```

**Repetition:** The `for` loop processes every score.

**Selection:** The `if/else` decides whether each score is a pass or fail.

## 8. Tracing an Algorithm

Consider:

```java
int[] numbers = {2, 7, 4, 9};

for (int number : numbers) {
    if (number % 2 == 0) {
        System.out.println(number + " is even");
    } else {
        System.out.println(number + " is odd");
    }
}
```

Trace it one item at a time:

| Number | `number % 2 == 0` | Output |
|---:|---|---|
| 2 | true | 2 is even |
| 7 | false | 7 is odd |
| 4 | true | 4 is even |
| 9 | false | 9 is odd |

**Tracing** means following an algorithm step by step to determine its output.

## 9. Practice: Selection

**Question:** What will this Java code print?

```java
int temperature = 80;

if (temperature > 90) {
    System.out.println("Hot");
} else if (temperature >= 70) {
    System.out.println("Warm");
} else {
    System.out.println("Cool");
}
```

**Answer:** `Warm`

The first condition is false. The second condition, `80 >= 70`, is true, so the `else if` block runs.

## 10. Practice: Repetition

**Question:** How many times does `Done` print?

```java
for (int i = 0; i < 3; i++) {
    System.out.println("Done");
}
```

**Answer:** 3 times.

The counter takes the values `0`, `1`, and `2`. Each value causes one iteration.

## 12. AP Computer Science Connection

Selection and repetition are important **control structures** for building algorithms.

When reading or writing Java code, ask:

1. **Where is the decision?** → Look for `if`, `else if`, and `else`.
2. **Where is the repetition?** → Look for `for` and `while`.
3. **What condition controls it?** → Identify the Boolean expression.
4. **How many times does it execute?** → Trace the loop carefully.
5. **What is the output?** → Follow the algorithm step by step.
